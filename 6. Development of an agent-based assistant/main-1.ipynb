```python
!pip install python-dotenv
!pip install crewai
from dotenv import load_dotenv
from openai import OpenAI
from crewai import Agent, Task, Crew
import os
```

```python
load_dotenv()
client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url=os.getenv("OPENAI_BASE_URL")
)
```

```python
# === 1. Researcher Agent ===
researcher = Agent(
    role="Researcher",
    goal="Gather detailed information about how attention works in neural networks.",
    backstory="You are an expert researcher who finds accurate and relevant information.",
    llm=client,
    verbose=True,
    allow_delegation=False,
    memory=True,
    system_prompt="""
You are Researcher Agent. Your task is to gather accurate, structured and relevant information about the topic: 
'How does attention mechanism work in neural networks?' 
Collect definitions, history, key concepts, types of attention, examples, and practical applications. 
Do not analyze — only gather and summarize facts.
"""
)

# === 2. Analyst Agent ===
analyst = Agent(
    role="Analyst",
    goal="Analyze and structure the information collected by the Researcher.",
    backstory="You are an analytical expert who can organize complex information.",
    llm=client,
    verbose=True,
    allow_delegation=False,
    memory=True,
    system_prompt="""
You are Analyst Agent. Your task is to analyze and structure the information gathered by the Researcher.
Extract key points, remove redundant data, logically group the material, and prepare a clear outline.
Do not write the final report — only structure and interpret the findings.
"""
)

# === 3. Writer Agent ===
writer = Agent(
    role="Writer",
    goal="Write a clear and coherent explanation of how attention works in neural networks.",
    backstory="You are a skilled technical writer.",
    llm=client,
    verbose=True,
    allow_delegation=False,
    memory=True,
    system_prompt="""
You are Writer Agent. Your task is to write a well-structured explanation based on the Analyst’s outline.
Write clearly, academically, and precisely. Do not add new facts.
"""
)
```

```python
research_task = Task(
    description=(
        "Gather detailed information about how the attention mechanism works in neural networks. "
        "Include definitions, history, key concepts, types of attention, practical examples, and applications. "
        "Do NOT analyze — only collect structured facts."
    ),
    agent=researcher,
    expected_output="A structured collection of raw factual information about the attention mechanism."
)


analysis_task = Task(
    description=(
        "Analyze the information provided by the Researcher. Identify key concepts, remove redundant details, "
        "logically group the ideas, and create a clear, structured outline explaining how attention works. "
        "Do NOT write the final explanation — only produce an outline."
    ),
    agent=analyst,
    expected_output="A logically structured outline explaining the attention mechanism in neural networks."
)


writing_task = Task(
    description=(
        "Write a clear, coherent, and well-structured explanation of how the attention mechanism works "
        "based on the Analyst’s outline. Use an academic yet accessible tone appropriate for a lab report. "
        "Do NOT introduce new facts — only elaborate on the provided outline."
    ),
    agent=writer,
    expected_output="A polished, well-written explanation suitable for a student's lab report."
)
```

```python
crew = Crew(
    agents=[researcher, analyst, writer],
    tasks=[research_task, analysis_task, writing_task],
    verbose=True
)

result = crew.kickoff()
print("\n=== FINAL RESULT ===\n")
print(result)
```
